# Importanto as Bibliotecas

```python
# https://docs.google.com/spreadsheets/d/1Ci5mEXBI5g0Q86dtDE-pLQpGuFUMfYwBzRStQeEb7mk/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1Ci5mEXBI5g0Q86dtDE-pLQpGuFUMfYwBzRStQeEb7mk/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

# Leitura Geral da Tabela do Sheets

```python
print("---- Dados Coletados ----")
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

# Iteração das respostas do formulário

```python
# iterar cada resposta do formulario
for indice, linha  in dados_planilha.iterrows():
  nome_local = linha["Nome do Local"]
  lat = linha["Latitude"]
  lon = linha["Longitude"]

  parametros = {
    'latitude':lat,
    'longitude':lon,
    'current':'temperature_2m'
    }


  print(f"---- Informações sobre {nome_local}")
  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print("\n\n")
```
